------> Outlier analysis identified some statistically extreme observations using the IQR method. However, the observed values were within their logical ranges, so legitimate observations were retained rather than removed.”

Feature Engineering

###Experience category (Fresher / Junior / Senior)

```python
df["experience_category"] = pd.cut(
    df["years_of_experience"],
    bins=[-1, 0, 3, float("inf")],
    labels=["Fresher", "Junior", "Senior"]
)
```

```python
print(df["experience_category"].value_counts())
```

##Academic performance bands




an academic performance band by calculating the average of SSC, HSC, and degree percentages and categorizing candidates into Low, Medium, and High academic performance groups

```python
df["academic_average"] = (
    df["ssc_percentage"] +
    df["hsc_percentage"] +
    df["degree_percentage"]
) / 3

df["academic_band"] = pd.cut(
    df["academic_average"],
    bins=[0,60,75,100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

print(df["academic_band"].value_counts())
```

Skills match level (Low / Medium / High)

converted the skills match percentage into three categorical levels—Low, Medium, and High—to make the feature easier to interpret and analyze

```python
df["skills_match_level"] = pd.cut(
    df["skills_match_percentage"],
    bins=[0, 50, 75, 100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

print(df["skills_match_level"].value_counts())
```

Interview performance category

an overall interview score by taking the average of technical, aptitude, and communication scores. I then categorized candidates into Low, Medium, and High interview performance levels.

```python
df["interview_score"]=(
    df["technical_score"]+
    df["aptitude_score"]+
    df["communication_score"]
)/3

df["interview_performance"] = pd.cut(
    df["interview_score"],
    bins=[0, 50, 75, 100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

print(df["interview_performance"].value_counts())
```
